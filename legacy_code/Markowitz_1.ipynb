```python
import yfinance as yf
import pandas as pd
import requests

def get_sp500_returns():
    # 1. Récupération de la liste des tickers du S&P 500 via Wikipédia
    url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

    # Ajout d'un User-Agent pour éviter le blocage 403 Forbidden
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"}
    response = requests.get(url, headers=headers)
    response.raise_for_status() # Lève une exception pour les codes d'état d'erreur

    table = pd.read_html(response.text)
    df_tickers = table[0]
    tickers = df_tickers['Symbol'].tolist()

    # Correction pour les tickers contenant des points (ex: BRK.B -> BRK-B pour Yahoo Finance)
    tickers = [t.replace('.', '-') for t in tickers]

    print(f"Téléchargement des données pour {len(tickers)} entreprises...")
    return (tickers)




liste= get_sp500_returns()
```

```python
len(liste)
tickers = liste
tickers[0]
```

```python
rendement=[]
variance=[]
i=0
for ticker in tickers:
  data= yf.download(ticker, start="2025-01-01", end="2026-01-01")["Close"]
  data.pct_change()[1:]
  e=data.mean().iloc[0]
  rendement.append(e)
  v=data.std().iloc[0]
  variance.append(v)

  print(i)
  i+=1
```

On a un probleme de minimisation sous contraintes, par exemple variance inferier à 0.5

---
On cherche max (w.R) avec les contraintes : norme de w = 1 , et (w.V) < 0.5


```python
import numpy as np
```

```python
R=rendement
V=variance
print(len(R))
print(len(V))
print(V[0])
```

```python
def portfeuille(var_max,R,V):
  R = np.array(R)/500 # Convert R to a NumPy array
  V = np.array(V) # Convert V to a NumPy array
  l=len(R)
  w=np.zeros(l)
  N=500
  a=0.00001
  for k in range(N):

    w = w+a*R
    for _ in range(10):
      w=np.maximum(w,0)
      #on projette sur le demi plan (w.V) < 0.5
      w= w - max(0, (np.dot(w,V)-var_max)/np.linalg.norm(V)**2)*V
      w=np.maximum(w,0)
      #on projette pour ||w|| = 1
      w=w/np.linalg.norm(w)
  print("norm",np.linalg.norm(w),"min",min(w))
  #return ("return",500*np.dot(w,R),"varaince :",np.dot(w,V))
  return(w)
```

```python
omega=portfeuille(30,R,V)
```

```python
##juste ppur tester
from scipy.optimize import minimize
var_max=30
# On cherche à MINIMISER - (w.R) pour MAXIMISER (w.R)
def objective(w, R):
    return -np.dot(w, R)

# Contraintes
cons = [
    {'type': 'eq', 'fun': lambda w: np.linalg.norm(w) - 1}, # Norme = 1
    {'type': 'ineq', 'fun': lambda w: var_max - np.dot(w, V)} # var_max - w.V >= 0
]

# Bornes (Positivité)
bounds = [(0, None) for _ in range(len(R))]

res = minimize(objective, x0=R/np.linalg.norm(R), args=(R,),
               method='SLSQP', constraints=cons, bounds=bounds)

print("Résultat Scipy :", -res.fun)
print("Variance Scipy :", np.dot(res.x, V))
```

On verifie que le w trouver avec notre code d'optimisation est proche de la solution donner par les methodes de la bibliotheque scipy considérée comme fiable
